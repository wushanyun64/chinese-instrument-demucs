## 3. Monitor training

Check `outputs/` for the latest experiment signature (SIG).
Metrics are logged by Dora and can be viewed with:
```bash
dora info <SIG>
```

````python
# Find the latest experiment
outputs = sorted(
    Path('outputs').glob('*'),
    key=lambda p: p.stat().st_mtime, reverse=True
)
if outputs:
    print(f"Latest experiment: {outputs[0].name}")
else:
    print("No experiments yet — start training first")
````
